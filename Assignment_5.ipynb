```python
#1)Import NumPy
import numpy as np
print("Assignment 5")

#2)create one dimantional array
print("Create one domentional array")
arr = np.array([10, 20, 30, 40, 50])
print("Array:", arr)
print("\n")

#3)Find Size and Dimension of an Array
print("Size and dimention of array")
print("Number of elements:", arr.size)
print("Dimension:", arr.ndim)
print("\n")

#4)Create 2 dimentional array
print("Create 2 dimentional array")
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
print("2-D Array:")
print(arr)
print("\n")

#3)Basic Array Operations
print("Basic operations on array")
a = np.array([10, 20, 30, 40, 50])
print("Array:", a)
print("Addition:", a + 5) #addition
print("Subtraction:", a - 5) #subtraction
print("Multiplication:", a * 2) #multiplication
print("Division:", a / 2) #division
print("\n")

#4) Array Indexing
print("Indexing of array")
print("First element:", a[0])
print("Third element:", a[2])
print("Last element:", a[-1])
print("\n")

#5)Array Slicing
print("Slicing of array")
print("First three elements:", a[0:3])
print("Elements from index 2:", a[2:])
print("\n")

#6)Mathematical and Statistical Operations
print("Mathematical and statistical operations")
print("Marks:", a)
print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Maximum:", np.max(a))
print("Minimum:", np.min(a))


```
