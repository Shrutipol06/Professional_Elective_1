```python
import numpy as np

# Create an array
a = np.array([1, 4, 9, 16, 25])
b = np.array([2, 3, 4, 5, 6])

print("Array A:", a)
print("Array B:", b)

# 1. Addition
print("Addition:", np.add(a, b))

# 2. Subtraction
print("Subtraction:", np.subtract(a, b))

# 3. Multiplication
print("Multiplication:", np.multiply(a, b))

# 4. Division
print("Division:", np.divide(a, b))

# 5. Floor Division
print("Floor Division:", np.floor_divide(a, b))

# 6. Modulus
print("Modulus:", np.mod(a, b))

# 7. Power
print("Power:", np.power(a, 2))

# 8. Square Root
print("Square Root:", np.sqrt(a))

# 9. Absolute Value
print("Absolute Value:", np.abs(-a))

# 10. Exponential
print("Exponential:", np.exp(a))

# 11. Natural Logarithm
print("Natural Log:", np.log(a))

# 12. Log Base 10
print("Log Base 10:", np.log10(a))

# 13. Sine
print("Sine:", np.sin(a))

# 14. Cosine
print("Cosine:", np.cos(a))

# 15. Tangent
print("Tangent:", np.tan(a))

# 16. Maximum
print("Maximum:", np.maximum(a, b))

# 17. Minimum
print("Minimum:", np.minimum(a, b))

# 18. Greater Than
print("Greater Than:", np.greater(a, b))

# 19. Less Than
print("Less Than:", np.less(a, b))

# 20. Equal
print("Equal:", np.equal(a, b))
```
