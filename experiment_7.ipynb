```python
import pandas as pd

# Create a DataFrame
data = {
    "Name": ["Amit", "Sneha", "Rahul", "Priya"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 75, 88]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# 1. Display a column
print("\nName Column:")
print(df["Name"])

# 2. Add a new column
df["Grade"] = ["A", "A+", "B", "A"]

print("\nAfter Adding Grade Column:")
print(df)

# 3. Add a new row
df.loc[len(df)] = ["Neha", 20, 92, "A+"]

print("\nAfter Adding New Row:")
print(df)

# 4. Update data
df.loc[0, "Marks"] = 89

print("\nAfter Updating Marks:")
print(df)

# 5. Delete a column
df = df.drop("Grade", axis=1)

print("\nAfter Deleting Grade Column:")
print(df)

# 6. Filter data
print("\nStudents with Marks greater than 85:")
print(df[df["Marks"] > 85])

# 7. Sort data
print("\nData Sorted by Marks:")
print(df.sort_values("Marks", ascending=False))

# 8. Calculate average
print("\nAverage Marks:")
print(df["Marks"].mean())

# 9. Find maximum marks
print("\nMaximum Marks:")
print(df["Marks"].max())

# 10. Find minimum marks
print("\nMinimum Marks:")
print(df["Marks"].min())
```
