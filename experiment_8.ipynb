```python
import matplotlib.pyplot as plt

# Data
students = ["Amit", "Sneha", "Rahul", "Priya", "Neha"]
marks = [75, 90, 65, 85, 95]

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 45, 50]



# 1. LINE GRAPH
plt.figure()

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid()

plt.show()


# 2. BAR GRAPH
plt.figure()

plt.bar(students, marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()



# 3. HORIZONTAL BAR GRAPH
plt.figure()

plt.barh(students, marks)

plt.title("Student Marks - Horizontal Bar Graph")
plt.xlabel("Marks")
plt.ylabel("Students")

plt.show()


# 4. PIE CHART
subjects = ["Maths", "Science", "English", "Computer"]
subject_marks = [80, 75, 90, 95]

plt.figure()

plt.pie(
    subject_marks,
    labels=subjects,
    autopct="%1.1f%%"
)

plt.title("Marks Distribution")

plt.show()


# 5. SCATTER PLOT
age = [18, 19, 20, 21, 22, 23]
height = [155, 160, 165, 168, 172, 175]

plt.figure()

plt.scatter(age, height)

plt.title("Age vs Height")
plt.xlabel("Age")
plt.ylabel("Height")

plt.show()


# 6. HISTOGRAM
marks_data = [45, 50, 55, 60, 65, 65, 70, 72,
              75, 75, 78, 80, 82, 85, 88, 90,
              92, 95, 98]

plt.figure()

plt.hist(marks_data, bins=5)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()


# 7. AREA GRAPH
plt.figure()

plt.fill_between(months, sales)

plt.title("Monthly Sales - Area Graph")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()


# 8. MULTIPLE LINE GRAPH
maths = [70, 80, 75, 90, 85]
science = [65, 75, 80, 85, 90]

plt.figure()

plt.plot(students, maths, marker="o", label="Maths")
plt.plot(students, science, marker="o", label="Science")

plt.title("Student Performance")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.legend()
plt.grid()

plt.show()


# 9. SUBPLOTS
plt.figure(figsize=(10, 6))

# Line graph
plt.subplot(2, 2, 1)
plt.plot(months, sales)
plt.title("Line Graph")

# Bar graph
plt.subplot(2, 2, 2)
plt.bar(students, marks)
plt.title("Bar Graph")

# Scatter plot
plt.subplot(2, 2, 3)
plt.scatter(age, height)
plt.title("Scatter Plot")

# Histogram
plt.subplot(2, 2, 4)
plt.hist(marks_data, bins=5)
plt.title("Histogram")

plt.tight_layout()

plt.show()
```
