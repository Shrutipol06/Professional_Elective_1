```python

import matplotlib.pyplot as plt

# Data
marks = [45, 50, 55, 60, 60, 65, 70, 75, 80, 85, 90]

# Create histogram
plt.hist(marks, bins=5, edgecolor='black')

# Add title and labels
plt.title("Histogram of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

# Display graph
plt.show()
```
